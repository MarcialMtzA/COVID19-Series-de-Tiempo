Vamos a contabilizar el número de fallecimientos por enfermedades respiratorias en Mexico durante el 2020
Eso incluye enfermedades como covid19, influenza y otras enfermedades respiratorias

```python
# Seleccionamos fecha de defunción y quitamos aquellos que no fallecieron

COVID20 = COVID20.loc[
    COVID20["FECHA_DEF"] != "9999-99-99",
    ["FECHA_DEF"]
]

# Visualizamos el resultado
COVID20.head(10)
```

```python
# Vemos el número de fallecidos por día

COVID20["DEF_POR_DIA"] = (
    COVID20.groupby("FECHA_DEF")["FECHA_DEF"]
    .transform("size")
)

# Visualizamos el resultado
COVID20.head(10)
```

```python
# Transformamos la fecha para mayor comodidad usando pandas

COVID20["fecha"] = pd.to_datetime(
    COVID20["FECHA_DEF"],
    format="%Y-%m-%d"
)

# Ordenamos los datos por fecha
COVID20 = COVID20.sort_values(by="fecha")

# Visualizamos el resultado
COVID20.head(10)
```

```python
# Eliminamos los registros duplicados
COVID20 = COVID20.drop_duplicates()

# Seleccionamos las variables de importancia
COVID20 = COVID20[["fecha", "DEF_POR_DIA"]]

# Visualizamos el resultado
COVID20.head(10)
```

```python
# Gráfica de defunciones por día
gf1, ax = plt.subplots(figsize=(10, 5))

ax.scatter(
    COVID20["fecha"],
    COVID20["DEF_POR_DIA"]
)

# Configuración del eje X
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=2))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

# Rotar las fechas
plt.xticks(rotation=45)

# Etiquetas
ax.set_xlabel("Fecha")
ax.set_ylabel("Defunciones por día")
ax.set_title("Defunciones diarias en México - 2020")

# Ajustar espacios
plt.tight_layout()

# Mostrar gráfica
plt.show()
```

```python
# Calculamos las defunciones mensuales de 2020

COVID2020 = COVID20.copy()

# Obtenemos el primer día de cada mes
COVID2020["mes"] = (
    COVID2020["fecha"].dt.to_period("M").dt.to_timestamp()
)

# Agrupamos por mes y sumamos las defunciones
COVID2020 = (
    COVID2020.groupby("mes", as_index=False)
    .agg(total_mensual=("DEF_POR_DIA", "sum"))
)

# Visualizamos el resultado
COVID2020.head(12)
```

```python
# Conservamos únicamente los meses anteriores a 2021

COVID2020 = COVID2020.loc[
    COVID2020["mes"] < "2021-01-01"
]

# Visualizamos el resultado
COVID2020
```

```python
# Calculamos las defunciones mensuales de 2021

COVID2021 = COVID21.copy()

# Obtenemos el primer día de cada mes
COVID2021["mes"] = (
    COVID2021["fecha"].dt.to_period("M").dt.to_timestamp()
)

# Agrupamos por mes y sumamos las defunciones
COVID2021 = (
    COVID2021.groupby("mes", as_index=False)
    .agg(total_mensual=("DEF_POR_DIA", "sum"))
)

# Visualizamos el resultado
COVID2021
```

```python
# Conservamos únicamente los meses anteriores a 2022

COVID2021 = COVID2021.loc[
    COVID2021["mes"] < "2022-01-01"
]

# Visualizamos el resultado
COVID2021
```

```python
# Calculamos las defunciones mensuales de 2022

COVID2022 = COVID22.copy()

# Obtenemos el primer día de cada mes
COVID2022["mes"] = (
    COVID2022["fecha"].dt.to_period("M").dt.to_timestamp()
)

# Agrupamos por mes y sumamos las defunciones
COVID2022 = (
    COVID2022.groupby("mes", as_index=False)
    .agg(total_mensual=("DEF_POR_DIA", "sum"))
)

# Visualizamos el resultado
COVID2022
```

```python
# Conservamos únicamente los meses anteriores a 2023

COVID2022 = COVID2022.loc[
    COVID2022["mes"] < "2023-01-01"
]

# Visualizamos el resultado
COVID2022
```

```python
# Calculamos las defunciones mensuales de 2023

COVID2023 = COVID23.copy()

# Obtenemos el primer día de cada mes
COVID2023["mes"] = (
    COVID2023["fecha"].dt.to_period("M").dt.to_timestamp()
)

# Agrupamos por mes y sumamos las defunciones
COVID2023 = (
    COVID2023.groupby("mes", as_index=False)
    .agg(total_mensual=("DEF_POR_DIA", "sum"))
)

# Visualizamos el resultado
COVID2023
```

```python
# Unimos las cuatro bases de datos

df = pd.concat(
    [COVID2020, COVID2021, COVID2022, COVID2023],
    axis=0,
    ignore_index=True
)

# Visualizamos el resultado
df
```

```python
# Gráfica de defunciones mensuales de 2020 a 2023
gf5, ax = plt.subplots(figsize=(12, 5))

ax.plot(df["mes"], df["total_mensual"])

# Ajustamos las fechas del eje X
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

plt.xticks(rotation=45)

# Etiquetas y título
ax.set_xlabel("Fecha")
ax.set_ylabel("Defunciones mensuales")
ax.set_title("Defunciones mensuales por COVID-19 en México (2020-2023)")

plt.tight_layout()
plt.show()
```

Pero nosotros queremos una serie de tiempo entonces:

```python
# Creamos la serie temporal mensual
ts_data = pd.Series(
    df["total_mensual"].to_numpy(),
    index=pd.date_range(
        start="2020-01-01",
        periods=len(df),
        freq="MS"
    ),
    name="total_mensual"
)

# Visualizamos la serie temporal
print(ts_data)
```

graficamos:

```python
# Graficamos la serie temporal
fig, ax = plt.subplots(figsize=(12, 5))

ax.plot(ts_data.index, ts_data.values)

# Configuramos las fechas del eje X
ax.xaxis.set_major_locator(mdates.MonthLocator(interval=3))
ax.xaxis.set_major_formatter(mdates.DateFormatter("%Y-%m"))

plt.xticks(rotation=45)

# Título y etiquetas
ax.set_title("Número de fallecimientos por enfermedades respiratorias")
ax.set_xlabel("Fecha")
ax.set_ylabel("Número de fallecimientos")

plt.tight_layout()
plt.show()
```
